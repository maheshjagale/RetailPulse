```python
# Handling missing descriptions
missing_description = df_clean[df_clean["Description"].isna()]
```

```python
missing_description[[
    "StockCode",
    "Description",
    "Quantity",
    "Price",
    "Customer ID",
    "Country"
]].head(20)
```

```python
'''Tells how many unique descriptions are associated with each stock code. If a stock code has multiple descriptions
 it may indicate inconsistent data entry or product variations.'''

description_per_stock = (
    df_clean
    .dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .nunique()
)
```

```python
description_per_stock.head()
```

```python
multi_description = description_per_stock[description_per_stock > 1]

print("StockCodes with multiple descriptions:", len(multi_description))
```

```python
multi_description.sort_values(ascending=False).head(10)
```

```python
df_clean[df_clean["StockCode"] == "20713"][
    ["StockCode", "Description", "Quantity", "Price"]
].drop_duplicates()
```

```python
df_clean.shape
```

```python
missing_stockcodes = missing_description["StockCode"].unique()

known_stockcodes = df_clean.loc[
    df_clean["Description"].notna(),
    "StockCode"
].unique()

recoverable = missing_description[
    missing_description["StockCode"].isin(known_stockcodes)
]

print("Missing descriptions:", len(missing_description))
print("Potentially recoverable:", len(recoverable))

print(
    "Potentially recoverable %:",
    round(len(recoverable) / len(missing_description) * 100, 2)
)
```

```python
unique_description_per_stock = (
    df_clean
    .dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .nunique()
)

reliable_stockcodes = unique_description_per_stock[
    unique_description_per_stock == 1
].index

print(
    "StockCodes with exactly one description:",
    len(reliable_stockcodes)
)
```

```python
description_mapping = (
    df_clean[
        df_clean["StockCode"].isin(reliable_stockcodes)
        & df_clean["Description"].notna()
    ]
    .drop_duplicates("StockCode")
    .set_index("StockCode")["Description"]
)
```

```python
df_clean["Description"] = (
    df_clean["Description"]
    .fillna(df_clean["StockCode"].map(description_mapping))
)
```

```python
print("Remaining missing descriptions:",
      df_clean["Description"].isna().sum())
```

```python
df_clean.dtypes
```

```python
df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"],
    errors="coerce"
)
```

```python
print(df_clean["InvoiceDate"].isna().sum())

# Its 0 so it means the date column is clean and has no missing values.
```

```python
df_clean["IsCancellation"] = (
    df_clean["Invoice"].astype(str).str.startswith("C")
)

df_clean["IsCancellation"].value_counts()
```

```python
# Creating a new column called revenue (quantity * price)
df_clean["Revenue"] = (
    df_clean["Quantity"] * df_clean["Price"]
)
```

```python
df_clean[
    ["Quantity", "Price", "Revenue"]
].head()
```

```python
# Categorizing transactions based on price

df_clean["TransactionType"] = "Normal Sale"

df_clean.loc[
    df_clean["Price"] < 0,
    "TransactionType"
] = "Bad Debt Adjustment"

df_clean.loc[
    df_clean["Price"] == 0,
    "TransactionType"
] = "Zero Price or Adjustment"

df_clean.loc[
    (df_clean["Quantity"] < 0) &
    (df_clean["Price"] > 0),
    "TransactionType"
] = "Return or Cancellation"
```

```python
df_clean["TransactionType"].value_counts()
```

```python
# Creating date features for EDA

df_clean["Year"] = df_clean["InvoiceDate"].dt.year
df_clean["Month"] = df_clean["InvoiceDate"].dt.month
df_clean["Day"] = df_clean["InvoiceDate"].dt.day
df_clean["DayOfWeek"] = df_clean["InvoiceDate"].dt.dayofweek

df_clean["YearMonth"] = (
    df_clean["InvoiceDate"].dt.to_period("M")
)

df_clean["Date"] = (
    df_clean["InvoiceDate"].dt.date
)
```
